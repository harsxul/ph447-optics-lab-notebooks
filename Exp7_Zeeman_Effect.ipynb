## Theory

In the initial Bohr atomic model, electrons orbit the nucleus in specific, quantized orbits, which correspond to discrete angular momentum values that are integer multiples of ℏ. The arrangement of electrons within these orbits determines the energy levels of atoms. When electrons transition between these orbits, the atom's energy changes accordingly. One way this quantized energy change is manifested is through the emission or absorption of photons. By analyzing the emitted photons—specifically their energy and polarization—we can infer the atom's energy levels and associated momenta.

The emission spectra of hydrogen atoms, for instance, are categorized into different series based on the initial and final energy levels of the atom. In the laboratory, we study these series to understand atomic transitions. The energy levels of atoms are sharp, and the differences between them generate photons of precise energies. These photon energies appear as sharp lines on a photographic plate when analyzed with a spectrograph. A spectrograph, similar to the prism spectrometer used in undergraduate labs, disperses light into its component wavelengths and records these as distinct bands or lines on a photographic plate. These emissions are known as "emission lines" and represent transitions between atomic energy levels rather than the energy levels themselves.

An interesting phenomenon related to emission lines is the Zeeman effect, which causes additional lines to appear when an external magnetic field is applied. Initially, Lorentz's semiclassical theory, which used the Bohr model and considered electrons as rotating charged particles generating magnetic moments, partially explained the Zeeman effect for some atoms like He, Zn, Ca, and Hg. However, this theory failed to account for the splitting of emission lines observed in atoms such as Na and Cr, where the original lines were omitted—this is known as the anomalous Zeeman effect.

The discovery of electron spin through experiments like Stern-Gerlach, coupled with the development of quantum mechanics, provided a more comprehensive explanation. Schrödinger’s quantum mechanical theory could explain the hydrogen emission lines, but it did not address the fine structure of hydrogen’s spectra. The incorporation of relativity and the interaction (or coupling) of spin and orbital angular momentum in quantum mechanics eventually provided a complete understanding of both the normal and anomalous Zeeman effects.

In summary, the mathematical relationship in Lorentz’s explanation connects the magnetic moment of an electron to its orbital angular momentum, forming a foundation for understanding atomic interactions under magnetic fields.

## Formulas

$$
\overrightarrow{\boldsymbol{u}}_J=g_J \frac{\mu_B}{h} \overrightarrow{\boldsymbol{J}}
$$
where $\overrightarrow{\boldsymbol{J}}=\overrightarrow{\boldsymbol{L}}+\overrightarrow{\boldsymbol{S}}$ and, the expression of new Lande factor $g_J$ is
$$
g_J=1+\frac{J(J+1)+S(S+1)-L(L+1)}{2 J(J+1)}
$$
The Lande factor $𝑔_𝐽$ can be calculated for the atomic level and energy variation can be calculated
as
$$\Delta E_J=g_J \vec{\mu}_B \cdot \vec{B} \text { or } \Delta E_J=g_J m_J \mu_B B $$
where $𝑚_𝐽$ is projection of $𝜇_𝐵$ on $𝐵$.

The Energy of photon going from ${ }^3 \mathrm{~S}_1 \mathrm{~m}_{\mathrm{J}}=+1$ to ${ }^3 \mathrm{P}_2 \mathrm{~m}_{\mathrm{J}}=+1$
$$
\left(E_H+\Delta E_H\right)-\left(E_L+\Delta E_L\right)
$$
putting $\Delta E$ from equation 2 and putting values of $m_J$
$$
\left(E_H-E_L\right)+\Delta E_H-\Delta E_L=\left(E_H-E_L\right)+g_{J H}(+1) \mu_B B-g_{J L}(+1) \mu_B B
$$

Similarly, for ${ }^3 \mathrm{~S}_1 \mathrm{~m}_{\mathrm{J}}=-1$ to ${ }^3 \mathrm{P}_2 \mathrm{~m}_{\mathrm{J}}=-1$
$$
\left(E_H-E_L\right)+g_{J H}(-1) \mu_B B-g_{J L}(-1) \mu_B B
$$

The energy difference between two transitions,
$$
\mu_B B\left(2 g_{J H}-2 g_{J L}\right)=2 \mu_B B\left(g_{J H}-g_{J L}\right)
$$

If the value of energy difference of two transition and $g_J$ for both atomic levels and magnetic field are known, the value of Bohr magneton can be calculated. In the experiment, we will use the same approach and find the Bohr magneton value for verification of above quantum theory.

$$ d = \frac{\lambda F^2}{\chi^2_{n-1} - \chi^2_n} $$
$$
|\vec{\mu_B}| = \frac{hc(\chi_{n+}^2 - \chi_{n-}^2)}{4(g_{JH} - g_{JL})\lambda F^2 |\vec{B}|}
$$


Selection rules are criteria which decide the allowed transitions between two energy levels. When $\Delta m_J=0, \pm 1$ the transitions are allowed. The value of $\Delta m_J$ and direction of emission decides the polarization. The emission from $\Delta m_J=0, \pm 1$ in perpendicular direction of magnetic field are plane polarized. The plane of polarization is different for 0 and $\pm 1$. The emission from $\Delta m_J= \pm 1$ will emit circularly polarized light perpendicular in the direction of magnetic field.

```python
import pandas as pd
import numpy as np

# Load the data into a DataFrame
data = pd.DataFrame({
    'Angle': [1, 2, 3, 4],
    'Length': [319.160, 425.367, 504.390, 575.271]
})

# Perform calculations
print("\n--- Performing Calculations ---")

# 1. Convert 'Length' to 'Radius (Micrometer)'
data['Radius (Micrometer)'] = data['Length'] * 3.9

# 2. Convert Radius from Micrometers to Meters
data['Radius (Meter)'] = data['Radius (Micrometer)'] * 1e-6

# 3. Calculate 'Radius² (Meter²)'
data['Radius² (Meter²)'] = data['Radius (Meter)'] ** 2

# 4. Calculate 'ΔR² (Meter²)'
data['ΔR² (Meter²)'] = data['Radius² (Meter²)'].shift(-1) - data['Radius² (Meter²)']

# 5. Constant
F = 0.085  # Focal length in meters

# 6. Formula for d: d = ΔR² / (4 * F)
data['d (Meters)'] = data['ΔR² (Meter²)'] / (4 * F)

# 7. Convert 'd' from Meters to Micrometers
data['d (Micrometers)'] = data['d (Meters)'] * 1e6

# 8. Handle NaN in the last row
data.at[len(data) - 1, 'd (Meters)'] = np.nan
data.at[len(data) - 1, 'd (Micrometers)'] = np.nan

# Display the final DataFrame
print("\nData with Calculations:")
print(data.to_string(index=False))

# Calculate and display the average 'd', excluding NaN
average_d = data['d (Micrometers)'].mean()
print(f"\nAverage d = {average_d:.6f} micrometers")
```

```python
import pandas as pd
import numpy as np

# Input your measured data
data = pd.DataFrame({
    'S.No': [1, 2, 3, 4],
    'Chi_plus_pixels': [325, 425, 510, 585],   # Replace with your measured Chi_n^+ values
    'Chi_minus_pixels': [295, 395, 485, 560]   # Replace with your measured Chi_n^- values
})

# Conversion factors
pixel_to_micrometer = 3.9         # μm per pixel
micrometer_to_meter = 1e-6        # m per μm

# Constants
h = 6.62607015e-34                # Planck's constant (J·s)
c = 2.99792458e8                  # Speed of light (m/s)
lambda_ = 546e-9                  # Wavelength (m)
F = 0.085                         # Focal length (m)
delta_g_J = 1                     # g_J^H + g_J^L = 1
B_Tesla = 0.50                    # Magnetic field in Tesla at 2.75 A (replace with actual value)

# Step 2: Convert pixel values to meters
data['Chi_plus_micrometers'] = data['Chi_plus_pixels'] * pixel_to_micrometer
data['Chi_minus_micrometers'] = data['Chi_minus_pixels'] * pixel_to_micrometer

data['Chi_plus_meters'] = data['Chi_plus_micrometers'] * micrometer_to_meter
data['Chi_minus_meters'] = data['Chi_minus_micrometers'] * micrometer_to_meter

# Step 3: Calculate Chi^2 values in meters squared
data['Chi_plus_squared_m2'] = data['Chi_plus_meters'] ** 2
data['Chi_minus_squared_m2'] = data['Chi_minus_meters'] ** 2

# Step 4: Compute Delta Chi^2
data['Delta_Chi_squared_m2'] = data['Chi_plus_squared_m2'] - data['Chi_minus_squared_m2']

# Step 5: Calculate mu_B using the refined formula
data['mu_B_J_per_G'] = (h * c * data['Delta_Chi_squared_m2']) / (4 * lambda_ * F**2 * B_Tesla)

# Accepted value of Bohr magneton in J/T
mu_B_theoretical_J_per_G = 9.2740100783e-24

# Calculate percentage error for each measurement
data['Percentage_Error'] = abs((data['mu_B_J_per_G'] - mu_B_theoretical_J_per_G) / mu_B_theoretical_J_per_G) * 100

# Calculate average mu_B and average percentage error
average_mu_B_J_per_G = data['mu_B_J_per_G'].mean()
average_percentage_error = data['Percentage_Error'].mean()

# Display the results
print("\nData with Calculations:")
print(data[['S.No', 'Chi_plus_pixels', 'Chi_minus_pixels', 'Delta_Chi_squared_m2', 'mu_B_J_per_G', 'Percentage_Error']])

print(f"\nAverage μB = {average_mu_B_J_per_G:.5e} J/G")
print(f"Theoretical μB = {mu_B_theoretical_J_per_G:.5e} J/G")
print(f"Average Percentage Error = {average_percentage_error:.2f}%")
```
