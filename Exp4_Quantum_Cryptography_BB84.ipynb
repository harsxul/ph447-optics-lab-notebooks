## 3. Theory

Cryptography is the practice of securing communication, ensuring that only authorized parties can access the intended message. Classical cryptographic techniques, though widely used, have vulnerabilities, particularly when subjected to brute-force attacks.

The BB84 protocol, proposed by Charles Bennett and Gilles Brassard in 1984, uses the principles of quantum mechanics to guarantee secure key distribution. The core advantage is that any interception of the message will disturb the quantum states, making eavesdropping detectable.

### **Key Principles:**
- **Polarization of Light**: Photons can be polarized in various directions, such as 0° (horizontal), 90° (vertical), 45°, and -45°. These polarization states can represent binary data (0s and 1s).
- **Quantum No-Cloning Theorem**: A fundamental principle in quantum mechanics stating that it is impossible to create an identical copy of an unknown quantum state.
- **BB84 Protocol**: This protocol uses two bases:
  - **+ Basis**: 0° and 90° represent bits 0 and 1.
  - **X Basis**: 45° and -45° represent bits 0 and 1.
  
  Alice (sender) sends random bits using a randomly chosen basis, and Bob (receiver) measures the bits using a random basis. After the transmission, Alice and Bob publicly compare their bases (without revealing the actual bit values). Only the bits transmitted and measured using the same basis are kept to generate the key.

### **Eavesdropper Detection**:
If an eavesdropper (Eve) tries to intercept the communication, she will disturb the quantum states, introducing detectable errors. Alice and Bob can detect her presence by publicly comparing a portion of their shared key.

## 4. Experimental Setup

- **Alice's Side (Sender)**: The laser diode emits light that passes through the half-wave plate, which rotates the polarization to one of four angles (0°, 90°, 45°, -45°). This polarized light represents the bit being transmitted.
- **Bob's Side (Receiver)**: Bob's half-wave plate randomly selects the basis (+ or X) for measurement. A polarizing beam-splitter directs the light to one of two detectors based on the polarization.
- **Eve's Setup (Optional)**: When Eve is added to the experiment, she tries to intercept the polarized light, measure it, and retransmit it to Bob. However, her random basis choice introduces errors.

## 5. Procedure

### 5.1 Key Generation
1. **Setup**: Alice and Bob are positioned facing each other, approximately 60 cm apart.
2. **Random Bit Selection**:
   - Alice generates two sets of random binary numbers (0s and 1s) for 52 bits in total. Bob generates one set of 52 random bits.
   - Alice randomly selects the polarization angles (0°, 90°, 45°, -45°) based on her bit values.
3. **Bit Transmission**: Alice sends the bits using her random basis. Bob measures the bits using his random basis.
4. **Basis Comparison**: Alice and Bob publicly announce the bases they used for each measurement. They discard any measurements where the bases do not match, keeping only the bits where the bases align. These bits form the encryption key.

### 5.2 Encryption and Transmission
1. **Encryption**: Alice selects a four-letter word to encrypt. She converts each letter into its binary form and adds the generated key using binary addition (XOR operation).
2. **Transmission**: Alice transmits the encrypted message by selecting a basis (+ or X) for each bit. Bob sets his apparatus to match the chosen basis and receives the encrypted bits.
3. **Decryption**: Bob uses the key to decrypt the received message by performing binary addition again. The result is the original word.

### 5.3 Eavesdropper Detection (Optional)
1. **Add Eve to the Setup**: Position Eve between Alice and Bob. Eve measures the light and forwards it to Bob. However, because Eve selects a random basis, her measurement introduces errors.
2. **Error Detection**: After the transmission, Alice and Bob compare a subset of their key bits. If the error rate exceeds 25%, they conclude that an eavesdropper is present.

## 1. Write in short about the working principle of the pulsed lasers.
Pulsed lasers emit light in short, controlled bursts, rather than continuously. These bursts, or pulses, can have very high peak power compared to continuous-wave lasers. The generation of pulses in lasers is often achieved by techniques such as **Q-switching** or **mode-locking**. In **Q-switching**, the laser's optical resonator is temporarily kept inactive, and energy is stored in the gain medium, which is then released as a short, intense pulse. In **mode-locking**, multiple longitudinal modes of the laser are phase-locked, causing constructive interference at regular intervals and leading to extremely short pulses (in the picosecond or femtosecond range). Pulsed lasers are useful for high-precision applications like material processing and medical treatments.

## 2. What is the material of the polarizing beam splitter? Explain its working principle.
Polarizing beam splitters (PBS) are typically made from **birefringent materials** such as calcite, quartz, or specialized glasses. The most common configuration uses a **polarizing cube beam splitter**, which consists of two right-angle prisms made of birefringent material, cemented together. A **thin dielectric coating** is applied at the hypotenuse where the prisms are joined.

### Working Principle:
The polarizing beam splitter works by splitting light into two beams with orthogonal polarization states. When unpolarized light enters the PBS, the coating at the interface reflects light with one polarization (usually the vertical polarization) and transmits the orthogonal polarization (horizontal polarization). As a result, the incoming light is split into two separate beams based on their polarization state. This is essential for quantum key distribution experiments like BB84, where polarization states carry the encoded information.

## 3. Show that the polarization vectors in one particular basis can be written as a linear combination of vectors in another basis. What is the probability that a 0° polarized photon passes through a 45° oriented polarizer?

### Linear Combination of Polarization Vectors:
Let’s consider the polarization vectors in the **+ basis** (0° and 90°) and the **X basis** (45° and -45°). The relationship between the polarization vectors in these bases can be expressed as:

- The 45° polarization state |45°⟩ can be written as:
  $[
  |45°⟩ = \frac{1}{\sqrt{2}} |0°⟩ + \frac{1}{\sqrt{2}} |90°⟩
  ]$
  
- The -45° polarization state |−45°⟩ can be written as:
  $[
  |−45°⟩ = \frac{1}{\sqrt{2}} |0°⟩ - \frac{1}{\sqrt{2}} |90°⟩
  ]$

Thus, polarization vectors in one basis can always be written as a linear combination of vectors in another basis.

### Probability Calculation:
The probability that a photon polarized at 0° passes through a 45° oriented polarizer is given by the square of the absolute value of the inner product of the 0° and 45° states:

$[
P(0° → 45°) = |\langle 45° | 0° \rangle|^2
]$

Using the linear combination relationship:

$[
\langle 45° | 0° \rangle = \frac{1}{\sqrt{2}}
]$

Thus, the probability is:

$[
P(0° → 45°) = \left( \frac{1}{\sqrt{2}} \right)^2 = \frac{1}{2} = 0.5
]$

Therefore, the probability that a 0° polarized photon passes through a 45° oriented polarizer is **50%**.

## 5. If the polarization of light falling on the half-wave plate makes an angle of 22.5° with the optic axis of the half-wave plate, this light after passing through the half-wave plate falls on the polarizing beam splitter. What would be the difference between the transmitted and reflected light in terms of intensity after passing through the polarizing beam splitter?

A half-wave plate (λ/2 plate) rotates the polarization of incident light by twice the angle between the light’s initial polarization and the optic axis of the wave plate.

### Step-by-Step Analysis:
- The incident light is polarized at 22.5° with respect to the optic axis of the half-wave plate.
- The half-wave plate rotates the polarization by:
  $[
  2 \times 22.5° = 45°
  ]$

After passing through the half-wave plate, the light is polarized at 45°.

Now, when the light falls on the polarizing beam splitter (PBS):
- A PBS transmits horizontally polarized light (0°) and reflects vertically polarized light (90°).
- For light polarized at 45°, it has an equal probability of being transmitted or reflected, as it is an equal superposition of 0° and 90° components.

### Intensity Calculation:
The transmitted intensity $(I_T)$ and reflected intensity $(I_R)$ for light polarized at 45° are both 50% of the total incident intensity $(I_0)$, because of the equal superposition:

$[
I_T = I_R = \frac{1}{2} I_0
]$

Thus, the transmitted and reflected intensities are equal, each being **50% of the incident light intensity**.
