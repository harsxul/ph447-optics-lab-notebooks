##**Theory:**
In this experiment, a lens is used to perform fourier transform and we demonstrate how it can be used in image editing using optical filtering.

A two-dimensional Fourier transformation $F$ of function $f$ is given by:
\begin{equation}
    F(u,v) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} f(x,y) e^{-\iota 2\pi (ux+vy)} dx dy
\end{equation}
while in inverse fourier transform is given by:
\begin{equation}
    f(x,y) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} F(u,v) e^{\iota 2\pi (ux+vy)} du dv
\end{equation}

The coefficients in $F(u,v)$ are generally complex numbers. Accordingly, they contain a real and an
imaginary component, or amplitude and phase information, that is
\begin{equation}
    F(u,v) = F_r(u,v) + \iota F_i(u,v) = |F(u,v)| e^{\iota \phi(u,v)}
\end{equation}
with the amplitude $|F(u,v)|$ and the phase $\phi(u,v) = \tan^{-1} \frac{F_i(u,v)}{F_r(u,v)}$.

When we examine the Fourier plane behind the objective lens, the human eye perceives the amplitude
but not the phase. This means that the brightness distribution alone does not contain the complete
information of the diffracting structure.

The process of image formation (with or without magnification) in terms of a Fourier transform can
be understood in following steps:

### **1. Fraunhofer diffraction as Fourier transform.**

Consider a plane wave travelling along the z-axis and incident at $z = 0$ on an object (transparency) with amplitude transmission function $f(x,y)$. The diffracted light is collected by a lens of focal length $F$ situated at $z=U$.

All light waves leaving the object in a particular angle are focused by the lens to a single point in the focal plane. The amplitude of the light at
that point in the focal plane, is therefore the sum of the amplitudes at points on the object plane, each multiplied with their phase
$\exp(\iota k_o \times optical\; path)$.

Therefore, we see that Fraunhofer diffraction pattern amplitude is given by the two-dimensional Fourier transform of the object transmission function $f(x,y)$.

In experiments, we usually measure the intensity, i.e.

$$ |\psi(u,v)|^2 = \int \int |f(x,y) e^{-\iota (ux+vy)} dx dy|^2 $$

From the above equation, we see that the information about phase is lost. There are ways to retrieve it.
However, when the object is at the focal plane before the lens, we obtain a true image.

### **2. Image formation as two Fraunhofer diffraction processes occurring sequentially**

Consider the object to be a diffraction grating. As we have seen above, at focal plane, we get the diffraction pattern which can be represented as a Fourier transform. From the focal plane, the waves overlap again at image plane, forming the image. The diffraction orders behave like a set of equally spaced point sources and the image is their interference pattern. Thus, the process of image formation appears to consist of two diffraction processes, applied sequentially.

The finest detail observable in the image is determined by the highest order of diffraction that is transmitted by the lens. In other words, the greater the extent of the Fraunhofer diffraction accessed by the imaging lens, the better the resolution. This is achieved by using as large an angular aperture as possible.

### Optical Spatial Filtering:

When representing a structure or an image with the help of a lens, high spatial frequencies (far away from the optical axis) can be assigned to fine structures and low spatial frequencies (close to the optical axis) to coarse structures. Optical filtering takes advantage of the Fourier transformation that takes place in the focal plane of a lens by manipulating the Fourier transform of an image rather than the image itself.

A simple lens setup demonstrates this principle. An object placed near the focal plane of a lens generates a real image on a screen, and an optical filter placed at the focal plane on the other side of the lens manipulates the Fourier transform of the image.

download (5).png

Optical filters are usually diaphragms intended to block specific information in the Fourier plane to generate a desired effect in the image, for example to soft-focus or sharpen an image. Information changes in the Fourier plane appear directly on the screen. Such a setup with a single lens is called a
2f setup and is represented schematically in Figure A. The reverse transformation from the Fourier plane to the image in the 2f setup takes place through the propagation of light from the focal plane of the lens to the screen.
To avoid large distances between the object plane and image plane, we can expand the 2f setup into a 4f setup which results in a shorter optical path. As shown in Figure B, the object plane in the 4f setup is equivalent to the front focal plane of the first lens. The rear focal plane of the first lens and the front focal plane of the second lens form the Fourier plane. The image plane is in the rear focal plane of the second lens. The reverse transformation from the Fourier plane is carried out by the second lens.

### Köhler illumination:

If one wants to examine an object with a microscope, one generally has the problem that the illuminating light source itself has a certain extent and structure (e.g. coiled filament, LED chip), which leads to uneven illumination of the object, may expose the object to heat, and produces scattered light. Köhler illumination, named after its developer August Köhler, allows one to avoid these issues and is common practice in microscopy today. The lighting system essentially consists of a collector lens for the light source, a condenser lens, and two diaphragms, the aperture diaphragm, and the field diaphragm.

## Loading and Displaying Images

In this section, we load the camera images and their corresponding Fourier transform images for two lattice structures. The images are displayed in grayscale to facilitate the analysis.


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage.color import rgb2gray

# Load and display the images for two lattice structures (High-Pass Filter)
camera_image_lattice1 = imread('/content/close_open_lattice.jpg')  # Replace with your actual file name
fourier_image_lattice1 = imread('/content/close-close_ft.jpeg')  # Replace with your actual file name

camera_image_lattice2 = imread('/content/lattice 2 camera img.jpg')  # Replace with your actual file name
fourier_image_lattice2 = imread('/content/lattice 2 fs.jpeg')  # Replace with your actual file name

# Convert to grayscale
camera_image_lattice1_grey = rgb2gray(camera_image_lattice1)
fourier_image_lattice1_grey = rgb2gray(fourier_image_lattice1)
camera_image_lattice2_grey = rgb2gray(camera_image_lattice2)
fourier_image_lattice2_grey = rgb2gray(fourier_image_lattice2)

# Display images
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].imshow(camera_image_lattice1_grey, cmap='gray')
ax[0, 0].set_title('Camera Image (Lattice 1)')
ax[0, 0].axis('off')

ax[0, 1].imshow(fourier_image_lattice1_grey, cmap='gray')
ax[0, 1].set_title('Fourier Image (Lattice 1)')
ax[0, 1].axis('off')

ax[1, 0].imshow(camera_image_lattice2_grey, cmap='gray')
ax[1, 0].set_title('Camera Image (Lattice 2)')
ax[1, 0].axis('off')

ax[1, 1].imshow(fourier_image_lattice2_grey, cmap='gray')
ax[1, 1].set_title('Fourier Image (Lattice 2)')
ax[1, 1].axis('off')

plt.show()
```

## Discrete Fourier Transform (DFT)

We compute the Discrete Fourier Transform (DFT) of the grayscale camera images using Python's FFT functions. The DFT represents the frequency domain of the image, revealing the spatial frequencies present in the lattice structures.


```python
# Perform DFT on camera images
dft_lattice1 = np.fft.fftshift(np.fft.fft2(camera_image_lattice1_grey))
dft_lattice2 = np.fft.fftshift(np.fft.fft2(camera_image_lattice2_grey))

# Display DFT results
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(np.log(np.abs(dft_lattice1)), cmap='gray')
ax[0].set_title('Discrete Fourier Transform (Lattice 1)')
ax[0].axis('off')

ax[1].imshow(np.log(np.abs(dft_lattice2)), cmap='gray')
ax[1].set_title('Discrete Fourier Transform (Lattice 2)')
ax[1].axis('off')

plt.show()
```

## Comparison of Experimental and Calculated Fourier Images

Finally, we compare the Fourier images obtained experimentally with those calculated using the Discrete Fourier Transform (DFT) method. This comparison helps us understand the similarities and differences between optical and mathematical Fourier analysis.


```python
# Display comparison of experimental and calculated Fourier images
fig, ax = plt.subplots(2, 2, figsize=(12, 10))

ax[0, 0].imshow(np.log(np.abs(dft_lattice1)), cmap='gray')
ax[0, 0].set_title('Calculated Fourier Transform (Lattice 1)')
ax[0, 0].axis('off')

ax[0, 1].imshow(fourier_image_lattice1_grey, cmap='gray')
ax[0, 1].set_title('Experimental Fourier Image (Lattice 1)')
ax[0, 1].axis('off')

ax[1, 0].imshow(np.log(np.abs(dft_lattice2)), cmap='gray')
ax[1, 0].set_title('Calculated Fourier Transform (Lattice 2)')
ax[1, 0].axis('off')

ax[1, 1].imshow(fourier_image_lattice2_grey, cmap='gray')
ax[1, 1].set_title('Experimental Fourier Image (Lattice 2)')
ax[1, 1].axis('off')

plt.show()
```

## High-Pass Filtering

High-pass filters are applied to the Fourier transform of the lattice images to observe how removing low-frequency components affects the image. The results show the frequency domain representation after filtering and the reconstructed spatial domain images.


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage.color import rgb2gray

def fourier_masker(image_path_camera, image_path_fourier, filter_radius):
    """
    Apply a high-pass filter to the Fourier transform of an image and compare it with the optical Fourier image.

    Parameters:
    - image_path_camera: Path to the camera image (e.g., Lattice 1 or Lattice 2 camera image)
    - image_path_fourier: Path to the optical Fourier image
    - filter_radius: Radius of the high-pass filter
    """

    f_size = 12

    # Load images
    optical_fourier_image = imread(image_path_fourier)
    optical_camera_image = imread(image_path_camera)

    # Convert to grayscale
    camera_image_grey = rgb2gray(optical_camera_image)

    # Compute Fourier Transform of the grayscale image
    camera_image_grey_fourier = np.fft.fftshift(np.fft.fft2(camera_image_grey))

    # Apply high-pass filter
    rows, cols = camera_image_grey_fourier.shape
    center_row, center_col = rows // 2, cols // 2

    # Create high-pass filter mask
    for x in range(rows):
        for y in range(cols):
            if ((x - center_row) ** 2 + (y - center_col) ** 2 <= filter_radius ** 2):
                camera_image_grey_fourier[x, y] = 0

    # Display results
    fig, ax = plt.subplots(2, 2, figsize=(10, 8))

    ax[0, 0].imshow(np.log(abs(camera_image_grey_fourier)), cmap='gray')
    ax[0, 0].set_title('Masked Discrete Fourier', fontsize=f_size)

    ax[0, 1].imshow(abs(np.fft.ifft2(np.fft.ifftshift(camera_image_grey_fourier))),
                    cmap='gray')
    ax[0, 1].set_title('Transformed Grayscale Image', fontsize=f_size)

    ax[1, 0].imshow(rgb2gray(optical_fourier_image), cmap='gray')
    ax[1, 0].set_title('Optical Fourier Image', fontsize=f_size)

    ax[1, 1].imshow(camera_image_grey, cmap='gray')
    ax[1, 1].set_title('Camera Image', fontsize=f_size)

    fig.suptitle("High Pass Filter", y=0.95)
    plt.show()

# Paths to images
lattice1_camera_image_path = '/content/lattice 2 camera img.jpg'  # Replace with actual image path
lattice1_fourier_image_path = '/content/lattice 2 fs.jpeg'  # Replace with actual image path

lattice2_camera_image_path = '/content/final high pass 2nd lattice.jpg'  # Replace with actual image path
lattice2_fourier_image_path = '/content/high pass filter L2.jpeg'  # Replace with actual image path

# Apply high-pass filter to Lattice 1 images
fourier_masker(lattice1_camera_image_path, lattice1_fourier_image_path, filter_radius=100)

# Apply high-pass filter to Lattice 2 images
fourier_masker(lattice2_camera_image_path, lattice2_fourier_image_path, filter_radius=100)
```

## Spatial Filtering

Spatial filters are applied along specific orientations (X and Y axes) in the Fourier domain. This section demonstrates how filtering out certain spatial frequencies can manipulate the resulting images, providing insights into the structure of the lattice images.


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage.color import rgb2gray

def spatial_filter_analysis(image_path_camera, image_path_fourier, lattice_name, filter_type='x'):
    """
    Apply a spatial filter to the Fourier transform of an image along the x-axis or y-axis.

    Parameters:
    - image_path_camera: Path to the camera image (e.g., Lattice 1 or Lattice 2 camera image)
    - image_path_fourier: Path to the optical Fourier image
    - lattice_name: Name of the lattice ('Lattice 1' or 'Lattice 2')
    - filter_type: Type of spatial filter to apply ('x' for x-axis filter, 'y' for y-axis filter)
    """

    f_size = 12

    # Load images
    optical_fourier_image = imread(image_path_fourier)
    optical_camera_image = imread(image_path_camera)

    # Convert to grayscale
    camera_image_grey = rgb2gray(optical_camera_image)

    # Compute Fourier Transform of the grayscale image
    camera_image_grey_fourier = np.fft.fftshift(np.fft.fft2(camera_image_grey))

    # Apply spatial filter
    rows, cols = camera_image_grey_fourier.shape

    if filter_type == 'x':
        # Spatial filter along x-axis
        camera_image_grey_fourier[0:rows//2 - 10, cols//2-5:cols//2+5] = 0
        camera_image_grey_fourier[rows//2 + 10:, cols//2-5:cols//2+5] = 0
        title_suffix = "Spatial Filter along X Axis"
    elif filter_type == 'y':
        # Spatial filter along y-axis
        camera_image_grey_fourier[rows//2-5:rows//2+5, 0:cols//2-10] = 0
        camera_image_grey_fourier[rows//2-5:rows//2+5, cols//2 + 10:] = 0
        title_suffix = "Spatial Filter along Y Axis"
    else:
        raise ValueError("Invalid filter type. Choose 'x' or 'y'.")

    # Display results
    fig, ax = plt.subplots(2, 2, figsize=(10, 8))

    ax[0, 0].imshow(np.log(abs(camera_image_grey_fourier)), cmap='gray')
    ax[0, 0].set_title(f'{lattice_name} Masked Discrete Fourier ({filter_type}-axis)', fontsize=f_size)

    ax[0, 1].imshow(abs(np.fft.ifft2(np.fft.ifftshift(camera_image_grey_fourier))),
                    cmap='gray')
    ax[0, 1].set_title(f'{lattice_name} Transformed Grayscale Image', fontsize=f_size)

    ax[1, 0].imshow(rgb2gray(optical_fourier_image), cmap='gray')
    ax[1, 0].set_title(f'{lattice_name} Optical Fourier Image', fontsize=f_size)

    ax[1, 1].imshow(camera_image_grey, cmap='gray')
    ax[1, 1].set_title(f'{lattice_name} Camera Image', fontsize=f_size)

    fig.suptitle(f'{lattice_name} - {title_suffix}', y=0.95)
    plt.show()


# Paths to images for Lattice 1 (adjust paths as needed)
lattice1_camera_image_horizontal_path = '/content/Spatial Filter 2.jpg'
lattice1_fourier_image_horizontal_path = '/content/lattice 2 furrior.jpeg'
lattice1_camera_image_vertical_path = '/content/Spatial Filter 1.jpg'
lattice1_fourier_image_vertical_path = '/content/Lattice 1 Vertical Fourrier.jpeg'

# Paths to images for Lattice 2 (adjust paths as needed)
lattice2_camera_image_horizontal_path = '/content/horizontal spatial 2 nd lattice.jpg'
lattice2_fourier_image_horizontal_path = '/content/lattice 2 horizontal furrier.jpeg'
lattice2_camera_image_vertical_path = '/content/vertical spatial 2 nd lattice.jpg'
lattice2_fourier_image_vertical_path = '/content/verticle lattice 2 furrier.jpeg'

# Perform spatial filter analysis for Lattice 1 - Horizontal Filter
spatial_filter_analysis(lattice1_camera_image_horizontal_path, lattice1_fourier_image_horizontal_path, 'Lattice 1', filter_type='x')

# Perform spatial filter analysis for Lattice 1 - Vertical Filter
spatial_filter_analysis(lattice1_camera_image_vertical_path, lattice1_fourier_image_vertical_path, 'Lattice 1', filter_type='y')

# Perform spatial filter analysis for Lattice 2 - Horizontal Filter
spatial_filter_analysis(lattice2_camera_image_horizontal_path, lattice2_fourier_image_horizontal_path, 'Lattice 2', filter_type='x')

# Perform spatial filter analysis for Lattice 2 - Vertical Filter
spatial_filter_analysis(lattice2_camera_image_vertical_path, lattice2_fourier_image_vertical_path, 'Lattice 2', filter_type='y')
```
