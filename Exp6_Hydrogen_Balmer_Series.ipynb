### Theory:

A hydrogen atom in a discharge lamp emits a series of lines in the visible part of the spectrum. This series is called the Balmer series, named after the Swiss teacher Johann Balmer, who in 1885, found by trial and error a formula to describe the wavelength of these lines. This formula is given by:

$$\frac{1}{\lambda} = R \left( \frac{1}{2^2} - \frac{1}{n^2} \right)$$

where $n$ are integers, 3, 4, 5, ... up to infinity, and $R$ is a constant now called the Rydberg constant.

Then in 1889, Johannes Robert Rydberg found several series of spectra that would fit a more general relationship, similar to Balmer's empirical formula. This general relation is known as the Rydberg formula and is given by:

$$\frac{1}{\lambda} = R \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right)$$

where $n_i > n_f$.

Here, $n_i$ and $n_f$ are integers, 1, 2, 3, 4, ... up to infinity, with $n_i > n_f$. For the hydrogen atom, $n_i = 2$ corresponds to the Balmer series.

### Spectrometer Calibration Using Hg Lamp:

Every element emits a characteristic spectrum of its own. To find the wavelength of light emitted, usually, the spectrum of a standard source is measured and compared. The unknown wavelength is determined by using Hartmann's equations. The wavelength of the prominent lines in any spectrum is studied by using Hartmann’s formula:

$$\lambda(d) = \lambda_0 + \frac{C}{d - d_0}$$

where $\lambda_0$, $C$, and $d_0$ are Hartmann’s constants that must be evaluated experimentally. This is done by substituting the distances $d$ of three known lines (in the mercury spectrum) from an arbitrary point and their wavelengths $\lambda$ into the above equation and solving the three equations. If we find three lines with scale readings $d_1$, $d_2$, and $d_3$ corresponding to known wavelengths $\lambda_1$, $\lambda_2$, and $\lambda_3$, respectively, then we can find Hartmann’s constants using the following relations:

$$A = \frac{\lambda_2 - \lambda_1}{\lambda_3 - \lambda_1}$$

$$B = \frac{d_2 - d_1}{d_3 - d_1}$$

$$\lambda_0 = \frac{A \lambda_3 - B \lambda_2}{A - B}$$

$$d_0 = \frac{B d_3 - A d_2}{B - A}$$

$$C = \frac{(\lambda_2 - \lambda_1) \cdot (d_0 - d_1) \cdot (d_0 - d_2)}{d_2 - d_1}$$

The wavelength of the prominent lines in any spectrum is then studied using Hartmann's formula:

$$\lambda(d) = \lambda_0 + \frac{C}{d - d_0}$$

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

# Cell 1: Plotting the Mercury Spectrum and Identifying Top 5 Peaks
hg_data = pd.read_excel("/content/Harshul_21B030015.xlsx")

# Plotting the mercury spectrum
plt.figure(figsize=(10, 6))
plt.plot(hg_data['Pixel'], hg_data['Raw Data'], color='blue', label='Hg Spectrum')
plt.xlabel('Pixel')
plt.ylabel('Intensity')
plt.title('Mercury Spectrum: Pixel vs Intensity')
plt.grid(True)
plt.legend()
plt.show()

# Identifying peaks in the Hg spectrum
peaks_hg, properties_hg = find_peaks(hg_data['Raw Data'], height=1000)  # Adjust height threshold if necessary
# Get the intensities of the identified peaks
peak_intensities_hg = hg_data['Raw Data'].iloc[peaks_hg]

# Sort by intensity and get top 5 peaks
top_5_hg = peak_intensities_hg.nlargest(5)
top_5_pixels_hg = hg_data['Pixel'].iloc[top_5_hg.index]

# Output the top 5 highest intensity peaks for Hg Spectrum
print("Top 5 Peaks for Hg Spectrum (Pixel vs Intensity):")
for pixel, intensity in zip(top_5_pixels_hg, top_5_hg):
    print(f"Pixel: {pixel}, Intensity: {intensity}")
```

### Step 1: Calculate the Wavelengths (λ)
We use the calibration equation to find λ for each pixel position:

**Given Hartmann Constants:**
- λ₀ = 254.26 nm
- d₀ = 5230.68 pixels
- C = 696459.23

#### For d₁ = 1362:
1. Compute d₀ - d₁:
   - d₀ - d₁ = 5230.68 - 1362 = 3868.68 pixels
2. Calculate λ₁:
   - λ₁ = 254.26 + (696459.23 / 3868.68)
   - λ₁ = 254.26 + 180.0388 = 434.2988 nm

#### For d₂ = 2229:
1. Compute d₀ - d₂:
   - d₀ - d₂ = 5230.68 - 2229 = 3001.68 pixels
2. Calculate λ₂:
   - λ₂ = 254.26 + (696459.23 / 3001.68)
   - λ₂ = 254.26 + 231.986 = 486.246 nm

#### For d₃ = 3487:
1. Compute d₀ - d₃:
   - d₀ - d₃ = 5230.68 - 3487 = 1743.68 pixels
2. Calculate λ₃:
   - λ₃ = 254.26 + (696459.23 / 1743.68)
   - λ₃ = 254.26 + 399.368 = 653.628 nm

**Calculated Wavelengths:**
- λ₁ = 434.299 nm
- λ₂ = 486.246 nm
- λ₃ = 653.628 nm

---

### Step 2: Calculate the Rydberg Constant (R)
The Balmer series formula is:

\[
\frac{1}{\lambda} = R \left( \frac{1}{2^2} - \frac{1}{n^2} \right)
\]

Rewriting for R:

\[
R = \frac{1}{\lambda} \left( \frac{1}{4} - \frac{1}{n^2} \right)^{-1}
\]

#### For λ₁ = 434.299 nm and n = 5:
1. Convert λ₁ to meters:
   - λ₁ = 434.299 × 10⁻⁹ m
2. Compute 1/λ₁:
   - 1/λ₁ = 1 / (434.299 × 10⁻⁹) = 2.301 × 10⁶ m⁻¹
3. Calculate the denominator:
   - (1/4) - (1/25) = (25 - 4) / 100 = 21 / 100
4. Compute R₁:
   - R₁ = 2.301 × 10⁶ × (100 / 21) = 10.949 × 10⁶ m⁻¹

#### For λ₂ = 486.246 nm and n = 4:
1. Convert λ₂ to meters:
   - λ₂ = 486.246 × 10⁻⁹ m
2. Compute 1/λ₂:
   - 1/λ₂ = 1 / (486.246 × 10⁻⁹) = 2.056 × 10⁶ m⁻¹
3. Calculate the denominator:
   - (1/4) - (1/16) = (16 - 4) / 16 = 3 / 16
4. Compute R₂:
   - R₂ = 2.056 × 10⁶ × (16 / 3) = 10.952 × 10⁶ m⁻¹

#### For λ₃ = 653.628 nm and n = 3:
1. Convert λ₃ to meters:
   - λ₃ = 653.628 × 10⁻⁹ m
2. Compute 1/λ₃:
   - 1/λ₃ = 1 / (653.628 × 10⁻⁹) = 1.530 × 10⁶ m⁻¹
3. Calculate the denominator:
   - (1/4) - (1/9) = (9 - 4) / 36 = 5 / 36
4. Compute R₃:
   - R₃ = 1.530 × 10⁶ × (36 / 5) = 11.016 × 10⁶ m⁻¹

---

### Step 3: Evaluate the Average Rydberg Constant (Rₐᵥg)
The calculated Rydberg constants are:
- R₁ = 10.949 × 10⁶ m⁻¹
- R₂ = 10.952 × 10⁶ m⁻¹
- R₃ = 11.016 × 10⁶ m⁻¹

**Average Rydberg Constant:**
\[
Rₐᵥg = \frac{R₁ + R₂ + R₃}{3} = 10.972 × 10⁶ m⁻¹
\]

Using the provided pixel positions and the calibration equation, we've calculated the wavelengths corresponding to the Balmer series transitions in hydrogen. Subsequently, we've evaluated the Rydberg constant \( R \) for each transition and found an average value:
