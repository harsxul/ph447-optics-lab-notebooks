### **Molecular Energy States**
A molecule has several electronic states, each comprising various vibrational and rotational states. Transitions between these states occur following selection rules.

Consider two electronic states: the lower (ground) state and the upper (first excited) state. The vibrational quantum number for the ground state is denoted by $υ''$ and for the upper state by $υ'$. Transitions between these states, denoted as $(υ', υ'')$, form vibrational bands. For example, the transition from $υ'=0$ to $υ''=0$ is called the $(0,0)$ band.

If visible light passes through an iodine vapor cell, the iodine molecules absorb the light, resulting in a band spectrum. This spectrum is due to electronic transitions between the ground and excited states. The $υ''=0$ to $υ'=0$ transition, or the $0-0$ transition, occurs at the lowest frequency (around 6200 Å). Discrete bands extend to about 5000 Å in the green region, beyond which a continuum is observed.

### **Total Energy of a Molecule**
The total energy $E$ of a molecule is the sum of its electronic E_e, vibrational E_v and rotational E_r energies:

$
E = E_e + E_v + E_r
$

In wave number units (cm⁻¹), this energy is represented as:

$$
T = T_e + G(υ) + F(J)
$$

Where $T_e$ $G(υ)$, and $F(J)$ are the term values for electronic, vibrational, and rotational energies, respectively.

### **Vibrational Energy Term Value**
The vibrational energy term value $G(υ)$ is given by:

$$
G(υ) = \frac{E_υ}{hc} = \left(υ + \frac{1}{2}\right)ω_e - \left(υ + \frac{1}{2}\right)^2 ω_e x_e + \left(υ + \frac{1}{2}\right)^3 ω_e y_e
$$

Where $ω_e$ is the frequency for infinitesimal vibration amplitudes, and $ω_e x_e$ and $ω_e y_e$ are anharmonicity constants.

### **Vibrational Energy Transitions**
The wave number $υ$ of a spectral line in a transition is determined by the difference between the term values of the upper ($T'$) and lower ($T''$) electronic states:

$$
υ = T' - T'' = \left(T_e' - T_e''\right) + \left(G'(υ') - G''(υ'')\right) + \left(F_υ(J') - F_υ(J'')\right)
$$

This can be simplified as:

$$
υ = υ_e + υ_υ + υ_r
$$

Where $υ_e$ represents the energy difference between the minima of the upper and lower electronic states.

### **Absorption Spectrum of Iodine**
The visible absorption spectrum of iodine results from the absorption of visible photons by iodine molecules, promoting electrons from the ground state to an excited state. The spectrum between 500-545 nm is due to transitions from the $υ''=0$ vibrational level in the ground state to various $υ'$ levels in the excited state. Beyond 545 nm, the spectrum becomes more complex due to transitions from $υ''=1$ and $υ''=2$ levels in the ground state.

Each peak in the spectrum corresponds to a transition between vibrational levels, forming the visible band system of iodine.

### **Vibrational Energy Term Values**
The vibrational energy term values are measured from the $υ=0$ vibrational level, not from the potential energy curve's minimum. The vibrational term value $G_0(υ)$ is given by:

$$
G_0(υ) = ω_0 υ - x_0 ω_0 υ^2 + y_0 ω_0 υ^3
$$

The difference between successive levels in the excited state is represented by $\Delta G(υ)$, which is the wave-number separation between two successive vibrational levels in one electronic state:

$$
\Delta G(υ) = G(υ+1) - G(υ) = ω_0 - ω_0 x_0 - 2 ω_0 x_0 υ
$$

The dissociation energy $D_0$ of the iodine molecule is given by:

$$
D_0 = \frac{ω_0}{4 ω_0 x_0}
$$

The second difference $\Delta^2 G(υ)$ is defined as:

$$
\Delta^2 G(υ) = \Delta G(υ+1) - \Delta G(υ) = -2 ω_0 x_0
$$

### **Morse Potential Function**

The potential energy of a diatomic molecule can be described using the Morse potential function as:

$$
U(r) = D_0 \left[1 - \exp\left(-\beta (r - R_e)\right)\right]^2
$$

Here, $D_0$ is the dissociation energy, $R_e$ is the equilibrium intermolecular distance, and $\beta$ is the Morse anharmonicity coefficient (which is different from the anharmonicity constant).

To derive the expression for $\beta$, consider the relationship between $\beta$ and the force constant $k$:

$$
\beta = \sqrt{\frac{k}{2D_0}}
$$

This expression can be derived directly from the Morse potential function.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Importing the dataset from an Excel file
# Replace 'iodine.xlsx' with your actual Excel file name
dataset = pd.read_excel('/content/group_last.xlsx', names=['Wavelength (nm)', 'Intensity (a.u.)'], skiprows=6)

# Display the dataset to verify it's loaded correctly
print(dataset)
```

```python
plt.figure(figsize=(12, 8))
plt.plot(dataset['Wavelength (nm)'], dataset['Intensity (a.u.)'])
```

```python
# Filter the dataset to include only wavelengths between 500 nm and 580 nm
filtered_data = dataset[(dataset['Wavelength (nm)'] > 500) & (dataset['Wavelength (nm)'] < 580)]

# Import the necessary function to detect peaks
from scipy.signal import find_peaks

# Identify the peaks in the intensity data, setting conditions for minimum height and prominence
detected_peaks, properties = find_peaks(filtered_data['Intensity (a.u.)'], height=0.1, distance=5, prominence=0.1)

# Create a plot to visualize the data along with the detected peaks
plt.figure(figsize=(12, 8))
plt.plot(filtered_data['Wavelength (nm)'], filtered_data['Intensity (a.u.)'], label="Intensity")
plt.scatter(filtered_data['Wavelength (nm)'].iloc[detected_peaks],
            filtered_data['Intensity (a.u.)'].iloc[detected_peaks],
            color='red', marker='x', label="Detected Peaks")

# Add labels and a legend to the plot
plt.xlabel('Wavelength (nm)')
plt.ylabel('Intensity (a.u.)')
plt.title('Detected Peaks in Iodine Absorption Spectrum')
plt.legend()
plt.show()
```

```python
# Initialize lists for quantum numbers and other values
quantum_number = []

# Loop through the detected peaks and reverse the order for the quantum numbers
for idx in range(len(detected_peaks)):
    quantum_number.append(detected_peaks[len(detected_peaks) - idx - 1])

# Lists to hold delta G and delta 2G values
delta_g_values = []
delta_2g_values = []

# Extract wavelengths and wave numbers for a specific range of quantum numbers
wavelengths_selected = [filtered_data['Wavelength (nm)'].iloc[quantum_number[i]] for i in range(12, 28)]
wave_numbers_selected = [(10**7) / filtered_data['Wavelength (nm)'].iloc[quantum_number[i]] for i in range(12, 28)]

# Compute ΔG and Δ²G values based on the wave numbers
for i in range(len(wave_numbers_selected) - 1):
    delta_g_values.append(round(wave_numbers_selected[i+1] - wave_numbers_selected[i], 2))

for i in range(len(delta_g_values) - 1):
    delta_2g_values.append(round(delta_g_values[i+1] - delta_g_values[i], 2))

# Print column headers with a new column for intensity
print(f"{'Quantum No.':<12}{'Wavelength (nm)':<18}{'Intensity (a.u.)':<18}{'Wave Number (cm⁻¹)':<24}{'ΔG':<10}{'Δ²G':<10}")

# Select relevant quantum numbers and store them
selected_quantum_numbers = [quantum_number[i] for i in range(12, 28)]

# Range for quantum number values
quantum_range = range(12, 28)

# Loop to print out the values with corresponding intensities
for i in range(len(selected_quantum_numbers)):
    qn_value = quantum_range[i]
    wavelength_value = round(wavelengths_selected[i], 2)
    intensity_value = round(filtered_data['Intensity (a.u.)'].iloc[selected_quantum_numbers[i]], 2)
    wave_number_value = round(wave_numbers_selected[i], 2)

    delta_g_val = delta_g_values[i] if i < len(delta_g_values) else "--"
    delta_2g_val = delta_2g_values[i] if i < len(delta_2g_values) else "--"

    print(f"{qn_value:<12}{wavelength_value:<18}{intensity_value:<18}{wave_number_value:<24}{delta_g_val:<10}{delta_2g_val:<10}")
```

```python
# Importing the necessary function from scipy for curve fitting
from scipy.optimize import curve_fit

# Define a function that represents a linear relationship
def linear_fit(x_values, slope, intercept):
    return slope * x_values + intercept

# Apply curve fitting to the data using the linear_fit function
fitted_params, _ = curve_fit(linear_fit, quantum_range[:15], delta_g_values)

# Extract the slope (m) and intercept (c) from the fitted parameters
slope, intercept = fitted_params

# Plot the data points and the fitted line
plt.plot(quantum_range[:15], delta_g_values, 'ro', label='Data')
plt.plot(quantum_range[:15], linear_fit(np.array(quantum_range[:15]), slope, intercept), 'b-', label='Fitted Line')

# Display the plot with labels
plt.xlabel('Quantum Number')
plt.ylabel('ΔG')
plt.legend()
plt.show()

# Output the slope and intercept
print(f"Slope (Gradient): {slope}", f"Intercept: {intercept}")
```

```python
# Perform curve fitting using the defined linear function
optimized_params, _ = curve_fit(linear_fit, quantum_range[:14], delta_2g_values)

# Extract the slope and intercept from the fitted parameters
slope_value, intercept_value = optimized_params

# Plot the delta_2g data points along with the fitted linear line
plt.plot(quantum_range[:14], delta_2g_values, 'ro', label='Data Points')
plt.plot(quantum_range[:14], linear_fit(np.array(quantum_range[:14]), slope_value, intercept_value), 'b-', label='Fitted Line')

# Show the plot with a legend and axis labels
plt.xlabel('Quantum Number')
plt.ylabel('Δ²G')
plt.legend()
plt.show()

# Print out the gradient and intercept
print(f"Slope (Gradient): {slope_value}", f"Intercept: {intercept_value}")
```
